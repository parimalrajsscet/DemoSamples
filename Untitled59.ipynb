###**Tokenization**
It means breaking a large piece of text into smaller piece called tokens

```python
# Sentence Tokenization
import nltk
```

```python
nltk.download('punkt_tab') # rule book for nltk to perform NLP steps
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
text = 'Hello! How are you doing? I am Learning NLP. It is really easy'

sentence = sent_tokenize(text) # based on punctuation (. ! ?)
print(sentence)
```

```python
text = 'Hello! How are you doing? I am Learning NLP. It is really easy'
words = word_tokenize(text)
print(words)
```

**SpaCy**

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text)
for i, sent in enumerate(doc.sents, 1):
  print(f'Sentence {i}: {sent.text}')
```

**Stopword removal**

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stopwordData = set(stopwords.words('english'))
```

```python
stopwords.fileids()
```

```python
# through nltk
text = 'This is a nice place and it can be very peaceful'
word = word_tokenize(text.lower())
filterd = [w for w in word if w not in stopwordData]
filterd
```

```python
# thrugh spacy
doc = nlp(text)
for token in doc:
  print(f'{token.text:<10} {token.is_stop}')
```
