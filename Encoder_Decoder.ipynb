```python
# Seq2Seq problems

# the input and output both are sequences

# 1. translation ( eng- spanish, eng-hindi)
# 2. text summarization
# 3. chatbot conversation
# 4. image to text generation
# 5. speech recognition
# 6. text to image
# 7. transcription



# In all these problems the input and output sequence length can be different
# Therefore the RNN and LSTM models will be able to do the task but not that efficiently.
```

```python
# Encoder - Decoder

# break the task into two parts and therefore increases the efficiency as well

# just like you doing one task is goiing to be maybe 50% efficient
# but if your friend joins you along in the task the efficiency can be 80%
```

## import the lib

```python
import numpy as np
import tensorflow as tf
```

```python
data = [ ("i am happy","मैं खुश हूँ"),
         ("You are sad","आप दुखी हैं"),
         ("she is tired", "वह थक गया है"),
         ("we are hungry","हम भूखें है"),
         ("they are busy","वे व्यस्त हैं"),
         ("i am cold","मुझे ठंड लग रही है"),
         ("you are late","तुम देरी से आए हो"),
         ("she is happy", "वह खुश है"),
         ("we are ready","हम तैयार हैं")]
```

```python
eng_sen =[]
hin_sen = []

for sen in data:
  eng_sen.append(sen[0])
  hin_sen.append(sen[1])
```

```python
eng_sen
```

```python
hin_sen
```

## Add \<start> and \<end>

```python
hin_sen = ['<start>'+sen+"<end>" for sen in hin_sen]
```

```python
hin_sen
```

## Tokenisation

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
eng_tokenizer = Tokenizer()
hin_tokenizer = Tokenizer()
```

```python
eng_tokenizer.fit_on_texts(eng_sen)
hin_tokenizer.fit_on_texts(hin_sen)
```

```python
eng_seq = eng_tokenizer.texts_to_sequences(eng_sen)
hin_seq = hin_tokenizer.texts_to_sequences(hin_sen)
```

```python
eng_seq
```

```python
hin_seq
```

```python
print(eng_tokenizer.word_index)
```

```python
print(hin_tokenizer.word_index)
```

```python
print(eng_tokenizer.word_counts)
```

```python
print(hin_tokenizer.word_counts)
```

```python
eng_vocab_size = len(eng_tokenizer.word_index) + 1
hin_vocab_size = len(hin_tokenizer.word_index) + 1
```

```python
eng_vocab_size
```

```python
hin_vocab_size
```

## Padding

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
eng_padded = pad_sequences(eng_seq, padding = 'post')
hin_padded = pad_sequences(hin_seq, padding = 'post')
```

```python
print(eng_padded)
print("*"*30)
print(hin_padded)
```

```python
max_eng_len = eng_padded.shape[1]
max_hin_len = hin_padded.shape[1]
```

```python
print(max_eng_len, max_hin_len)
```

```python
eng_padded = pad_sequences(eng_seq, maxlen= max_hin_len, padding = 'post')
```

```python
eng_padded
```

## RNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
model = Sequential()

model.add(Embedding(input_dim = eng_vocab_size,
                    output_dim = 64,
                    input_length = max_eng_len ))

model.add(SimpleRNN(64, return_sequences=True))

model.add(Dense(hin_vocab_size, activation = 'softmax'))

model.compile(optimizer = 'adam',
              loss = 'sparse_categorical_crossentropy',
              metrics=['accuracy'])

# hin_padded = np.expand_dims(hin_padded, -1)

model.build(input_shape=(None,max_eng_len))
model.summary()
```

```python
eng_padded.shape
```

```python
hin_padded.shape
```

```python
model.fit(eng_padded, hin_padded,
          epochs = 100,
          batch_size=2)
```

```python
model.fit(eng_padded, hin_padded,
          epochs = 100,
          batch_size=2)
```

```python
pred = model.predict(eng_padded)
```

```python
pred_seq = np.argmax(pred, axis = -1)
print(pred_seq)
```

```python
for seq in pred_seq:
  sentence = []
  for idx in seq:
    if idx != 0:
      word = hin_tokenizer.index_word.get(idx, " ")
      sentence.append(word)
  print(" ".join(sentence))
```

#Encoder - Decoder

```python
# Teacher Forcing

# we give correct previous word model predict next word
```

```python
decoder_input = hin_padded[:,:-1]
decoder_output = hin_padded[:, 1:]


print(decoder_input)
print("*"*20)
print(decoder_output)
```
